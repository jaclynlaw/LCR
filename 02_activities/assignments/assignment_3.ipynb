```python
# Import standard libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

#### Load the Wine dataset:

Use scikit-learn to load the Wine dataset and convert it into a Pandas DataFrame.

#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i], fontsize=24)
            plt.ylabel(feature_names[j], fontsize=24)
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')

            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**
- Do you notice any patterns or relationships between the different features? How might these patterns help in distinguishing between different species?

> Yes, there appears to be positive linear correlations between some pairs of features, such as total_phenols x nonflavanoid_phenols, or alcohol x proline, while other features are high in 1 parameter but low in another (e.g. high in hue but low in color_intensity). These relationships may indicate that certain types of wines may have specific characteristics associated with it. For example, 1 species may be characterized by high levels of both high total_phenols and flavanoids but another species may have low high total_phenols and flavanoids, and that there may be several species of wines that share common characteristics (represented by each data point).

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

> Since clustering algorithms, such as K-means, are dependent on distance calculations between two points, any feature that has a much larger scale would dominate the calculation regardless of the feature's actual importance in determining the clusters. For example, magnesium is measured on a larger magnitude than malic_acid, but magnesium itself might not be the determining factor for a wine's species identification. Therefore standardization is necessary to put all features on the same scale.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...

# Fit KMeans
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(scaled_wine_df)

# Assigned cluster labels to original data
clustered_wine_data = wine_df.copy()
clustered_wine_data['Cluster'] = kmeans.labels_



# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

> A common method for determining the optimal number of clusters is the elbow method. The elbow method helps us select the number of clusters in a more empirical way to avoid distorting the results such as too few clusters or too many clusters, both of which would artificially create patterns in the data that might not be genuine.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...


mean_color_intensity = wine_df['color_intensity'].mean()

np.random.seed(123)

bootstrap_samples = [] # Empty list to store the bootstrap samples

for i in range(10000):
    sample = wine_df['color_intensity'].sample(frac=1, replace=True)
    bootstrap_samples.append(sample.mean())  

lower_bound = np.percentile(bootstrap_samples, 5)
upper_bound = np.percentile(bootstrap_samples, 95)


# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")

# Plot the histogram of the bootstrap distribution for my own reference
plt.figure(figsize=(10, 6))
plt.hist(bootstrap_samples, bins=20, edgecolor='black', alpha=0.7)

sd_bootstrap_means = np.std(bootstrap_samples)
print(f"SD = {sd_bootstrap_means}")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

> Here we have a sample of 178 different wines, and we can calculate the mean color intensity of these 178 wines. The average of our single wine data set may not be representative of more bottles of wine. We use bootstrapping in this context to help us estimate the average color intensity of wines and this helps use understand how much the average might vary if we were to be able to sample other, different bottles of wine beyond our set of 178.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

> The 90% confidence interval helps us determine how likely we are to see the calculated average. In this case, we are 90% confident that our interval captures the true mean of the total wine population, i.e. we are 90% confident that the true mean color intensity of all wines lies between 4.78 and 5.35.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

> My bootstrapped means are normally distributed and tightly centred (narrow CI; standard deviation of means = 0.17) around the true mean of the data set, 5.1, with few outliers. This suggests that the mean of the color intensity appears to be a stable and reliable estimate.
